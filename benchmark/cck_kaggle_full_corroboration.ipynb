```python
# ── All core logic inlined ────────────────────────────────────────────────────
from __future__ import annotations
import os, re, json, math, random, statistics, ast, glob
from collections import Counter
from fastembed import TextEmbedding

# ── BM25 ─────────────────────────────────────────────────────────────────────
_WORD = re.compile(r'[A-Za-z0-9_]+')
_STOP = set('the a an of to in on for and or is are was were be by with vs at as '
            'it its this that what which who whose how many each'.split())

def _tokenize(s):
    return [w.lower() for w in _WORD.findall(s) if w.lower() not in _STOP]

def chunk(text, size=400, overlap=50):
    words = text.split()
    out, i = [], 0
    while i < len(words):
        out.append(' '.join(words[i:i+size]))
        i += size - overlap
    return out

class BM25:
    def __init__(self, chunks, k1=1.5, b=0.75):
        self.chunks = chunks; self.k1 = k1; self.b = b; self.N = len(chunks)
        toks = [_tokenize(c) for c in chunks]
        self.avgdl = sum(len(t) for t in toks) / max(self.N, 1)
        self.tf = [Counter(t) for t in toks]
        df = Counter()
        for t in toks:
            for w in set(t): df[w] += 1
        self.idf = {w: math.log((self.N-df[w]+0.5)/(df[w]+0.5)+1) for w in df}
    def score(self, query, idx):
        dl = sum(self.tf[idx].values()); s = 0.0
        for w in _tokenize(query):
            if w not in self.idf: continue
            tf = self.tf[idx].get(w, 0)
            s += self.idf[w]*(tf*(self.k1+1))/(tf+self.k1*(1-self.b+self.b*dl/self.avgdl))
        return s
    def topk(self, query, k):
        return [self.chunks[i] for i in sorted(range(self.N),
                key=lambda i: self.score(query,i), reverse=True)[:k]]

# ── Semantic retriever ────────────────────────────────────────────────────────
def _cos(a, b):
    dot = sum(x*y for x,y in zip(a,b))
    na = math.sqrt(sum(x*x for x in a)); nb = math.sqrt(sum(y*y for y in b))
    return dot/(na*nb) if na and nb else 0.0

class SemanticRetriever:
    def __init__(self, chunks, model_name='BAAI/bge-small-en-v1.5', max_chunks=100000, batch=64):
        self.capped = len(chunks) > max_chunks
        if self.capped:
            step = len(chunks)/max_chunks
            chunks = [chunks[int(i*step)] for i in range(max_chunks)]
        self.chunks = chunks; self.n_indexed = len(chunks)
        self._model = TextEmbedding(model_name)
        emb = []
        for i in range(0, len(chunks), batch):
            emb.extend(self._model.embed(chunks[i:i+batch]))
        self._emb = emb
    def topk(self, query, k):
        if not self._emb: return []
        qv = list(self._model.embed([query]))[0]
        return [self.chunks[i] for i in sorted(range(len(self.chunks)),
                key=lambda i: _cos(qv, self._emb[i]), reverse=True)[:k]]

# ── Corpus + question miner ───────────────────────────────────────────────────
def load_repo_text(path):
    parts = []
    for f in glob.glob(os.path.join(path,'**','*.py'), recursive=True):
        try:
            if os.path.getsize(f) < 200_000:
                parts.append(open(f, encoding='utf-8', errors='ignore').read())
        except OSError: pass
    return '\n'.join(parts)

def mine_conceptual(path, n):
    qs = []
    for f in glob.glob(os.path.join(path,'**','*.py'), recursive=True):
        try:
            if os.path.getsize(f) > 200_000: continue
            tree = ast.parse(open(f, encoding='utf-8', errors='ignore').read())
        except Exception: continue
        for node in ast.walk(tree):
            if isinstance(node,(ast.FunctionDef,ast.AsyncFunctionDef,ast.ClassDef)):
                doc = ast.get_docstring(node)
                if not doc: continue
                name = node.name
                if len(name)<4 or name.startswith('_'): continue
                first = re.split(r'(?<=[.!?])\s', doc.strip())[0].strip()
                if len(first)<25 or len(first)>200: continue
                if name.lower() in first.lower(): continue
                qs.append({'q':f'Which function or class {first[0].lower()+first[1:]}','fact':name})
    random.Random(0).shuffle(qs)
    seen=set(); out=[]
    for it in qs:
        if it['fact'] in seen: continue
        seen.add(it['fact']); out.append(it)
        if len(out)>=n: break
    return out

def bootstrap_ci(hits, nb=1000):
    if not hits: return (None,None,None)
    rng=random.Random(0); N=len(hits); ms=[]
    for _ in range(nb):
        ms.append(sum(hits[rng.randrange(N)] for _ in range(N))/N)
    ms.sort()
    return (round(statistics.mean(hits),4),round(ms[int(0.025*nb)],4),round(ms[int(0.975*nb)],4))

# ── Metrics ───────────────────────────────────────────────────────────────────
def m1_substring(chunks_list, fact):
    return fact.lower() in '\n'.join(chunks_list).lower()

_defre_cache = {}
def m2_defline(chunks_list, fact):
    if fact not in _defre_cache:
        _defre_cache[fact] = re.compile(r'(?:def|class)\s+'+re.escape(fact)+r'\b')
    return any(_defre_cache[fact].search(c) for c in chunks_list)

print('Core logic loaded.')
```

```python
# Build corpus and questions ONCE -- reused across all sweeps
REPOS = {'requests':'/tmp/requests','click':'/tmp/click','black':'/tmp/black'}
N_PER_REPO = 80  # 80 x 3 = 240 total
BOOTSTRAP  = 1000
OUT        = '/kaggle/working'

print('Loading corpus...', flush=True)
corpus   = {n: load_repo_text(p)  for n,p in REPOS.items()}
chunks_m = {n: chunk(corpus[n])   for n in corpus}
bm25_m   = {n: BM25(chunks_m[n]) for n in corpus}

print('Mining questions...', flush=True)
Q_all = []
for n,p in REPOS.items():
    for it in mine_conceptual(p, N_PER_REPO):
        if it['fact'].lower() in corpus[n].lower():
            it['project']=n; Q_all.append(it)
print(f'N={len(Q_all)} questions across {list(REPOS.keys())}')
assert len(Q_all) >= 200, f'Expected ~240 questions, got {len(Q_all)} -- check repo clones'
```

```python
# ── Main sweep function (reuses pre-built corpus and questions) ───────────────
def run_sweep(label, Q, chunks_m, bm25_m, k, max_chunks,
              embed_model, sem_threshold, out_path, bootstrap=1000):
    print(f'\n{"="*60}', flush=True)
    print(f'{label}  k={k}  model={embed_model}  cap={max_chunks}  thresh={sem_threshold}')
    print(f'{"="*60}', flush=True)

    # Build semantic index at this cap
    sem_m = {n: SemanticRetriever(chunks_m[n], model_name=embed_model, max_chunks=max_chunks)
             for n in chunks_m}
    model_inst = sem_m[list(sem_m)[0]]._model

    # Precompute fact embeddings for m3
    facts    = [q['fact'] for q in Q]
    fact_emb = dict(zip(facts, model_inst.embed([f'definition of {f}' for f in facts])))

    def retrieve(arm, q):
        p = q['project']
        return bm25_m[p].topk(q['q'],k) if arm=='bm25' else sem_m[p].topk(q['q'],k)

    def score_all(arm):
        h1,h2,h3=[],[],[]
        for q in Q:
            ch=retrieve(arm,q); fact=q['fact']
            s1=m1_substring(ch,fact); s2=m2_defline(ch,fact)
            if s1: s3=True
            else:
                ce=list(model_inst.embed(ch)); fe=fact_emb[fact]
                s3=any(_cos(fe,c)>=sem_threshold for c in ce)
            h1.append(int(s1));h2.append(int(s2));h3.append(int(s3))
        return h1,h2,h3

    result = {'label':label,'N':len(Q),'k':k,'max_chunks':max_chunks,
              'embed_model':embed_model,'sem_threshold':sem_threshold,
              'repos':list(REPOS.keys()),
              'metrics_explained':{
                  'm1':'substring anywhere (favors lexical)',
                  'm2':'def/class defline (stricter, neither favored)',
                  'm3':f'substring OR semantic-credit cos>={sem_threshold} (favors semantic)'},
              'arms':{}}

    for arm in ('bm25','semantic'):
        h1,h2,h3=score_all(arm)
        result['arms'][arm]={
            'm1_substring': bootstrap_ci(h1,bootstrap),
            'm2_defline':   bootstrap_ci(h2,bootstrap),
            'm3_sem_credit':bootstrap_ci(h3,bootstrap)}

    # Print + CI overlap check
    print(f'{"metric":<18}{"bm25 [CI]":>32}{"semantic [CI]":>32}')
    for m in ('m1_substring','m2_defline','m3_sem_credit'):
        b=result['arms']['bm25'][m]; s=result['arms']['semantic'][m]
        gap=round(b[0]-s[0],4)
        overlap = b[1]<=s[2] and s[1]<=b[2]
        flag = '' if overlap else '  <-- INVESTIGATE'
        print(f'{m:<18}{f"{b[0]} [{b[1]},{b[2]}]":>32}{f"{s[0]} [{s[1]},{s[2]}]":>32}  gap={gap}{flag}')

    # BM25 flatness check (internal control)
    bm25_m1 = result['arms']['bm25']['m1_substring'][0]
    print(f'BM25 m1={bm25_m1}  [internal control: must be ~stable across all runs]')

    with open(out_path,'w') as f: json.dump(result,f,indent=1)
    print(f'wrote {out_path}')
    return result

print('Sweep function ready.')
```

```python
# ── Run all 8 sweeps ──────────────────────────────────────────────────────────
# Each sweep reuses the pre-built corpus and questions.
# Semantic index is rebuilt per sweep (different cap or model).

results = {}

# --- GROUP 1: Embedder robustness (full corpus, thresh=0.72) ---
# Proves gap disappears regardless of embedder size

results['bge_base_full'] = run_sweep(
    'bge-base full corpus', Q_all, chunks_m, bm25_m, k=6, max_chunks=100000,
    embed_model='BAAI/bge-base-en-v1.5', sem_threshold=0.72,
    out_path=f'{OUT}/SWEEP_bge_base_full.json')

# --- GROUP 2: Threshold sensitivity (bge-small, full corpus) ---
# Proves m3 conclusion is not driven by the cosine threshold choice

results['thresh_065'] = run_sweep(
    'bge-small thresh=0.65', Q_all, chunks_m, bm25_m, k=6, max_chunks=100000,
    embed_model='BAAI/bge-small-en-v1.5', sem_threshold=0.65,
    out_path=f'{OUT}/SWEEP_threshold_0.65.json')

results['thresh_080'] = run_sweep(
    'bge-small thresh=0.80', Q_all, chunks_m, bm25_m, k=6, max_chunks=100000,
    embed_model='BAAI/bge-small-en-v1.5', sem_threshold=0.80,
    out_path=f'{OUT}/SWEEP_threshold_0.80.json')

# --- GROUP 3: Cap sweep replication (bge-small, thresh=0.72) ---
# Replicates the N=240 Kaggle cap sweep from CURRENT_STATE.md
# BM25 must be flat -- proves internal control holds in this run

for cap in [1000, 3000, 100000]:
    results[f'cap_{cap}'] = run_sweep(
        f'cap={cap} replication', Q_all, chunks_m, bm25_m, k=6, max_chunks=cap,
        embed_model='BAAI/bge-small-en-v1.5', sem_threshold=0.72,
        out_path=f'{OUT}/SWEEP_cap_{cap}.json')

# --- GROUP 4: k sensitivity (bge-small, full corpus, thresh=0.72) ---
# Proves finding holds at different retrieval depths

results['k3_full'] = run_sweep(
    'k=3 full corpus', Q_all, chunks_m, bm25_m, k=3, max_chunks=100000,
    embed_model='BAAI/bge-small-en-v1.5', sem_threshold=0.72,
    out_path=f'{OUT}/SWEEP_k3_full.json')

results['k10_full'] = run_sweep(
    'k=10 full corpus', Q_all, chunks_m, bm25_m, k=10, max_chunks=100000,
    embed_model='BAAI/bge-small-en-v1.5', sem_threshold=0.72,
    out_path=f'{OUT}/SWEEP_k10_full.json')
```

```python
# ── Final summary + internal control verification ─────────────────────────────
print('\n' + '='*70)
print('FINAL SUMMARY -- BM25 internal control + CI overlap check')
print('='*70)

bm25_m1_values = []

for label, r in results.items():
    bm25_m1 = r['arms']['bm25']['m1_substring'][0]
    bm25_m1_values.append(bm25_m1)
    print(f'\n{r["label"]}  N={r["N"]}  k={r["k"]}  cap={r["max_chunks"]}  model={r["embed_model"]}')
    for m in ('m1_substring','m2_defline','m3_sem_credit'):
        b=r['arms']['bm25'][m]; s=r['arms']['semantic'][m]
        gap=round(b[0]-s[0],4)
        overlap = b[1]<=s[2] and s[1]<=b[2]
        flag = 'CI_overlap=YES' if overlap else 'CI_overlap=NO <-- INVESTIGATE'
        print(f'  {m:<22} BM25={b[0]}  sem={s[0]}  gap={gap}  {flag}')

print('\n' + '='*70)
print('INTERNAL CONTROL: BM25 m1 across all runs (must be stable)')
bm25_range = max(bm25_m1_values) - min(bm25_m1_values)
for label, v in zip(results.keys(), bm25_m1_values):
    print(f'  {label:<20} BM25 m1={v}')
print(f'Range: {round(bm25_range,4)}  {"PASS -- stable" if bm25_range < 0.05 else "FAIL -- investigate"}')

print('\nDownload all SWEEP_*.json from the Output tab and commit to')
print('/Users/pupa/Projects/consistent-context-kit/benchmark/')
```
